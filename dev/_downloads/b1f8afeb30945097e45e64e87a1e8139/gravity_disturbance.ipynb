```python
import ensaio
import xarray as xr

fname = ensaio.fetch_earth_gravity(version=1)
gravity = xr.load_dataarray(fname)
gravity
```

```python
 import pygmt

 # Needed so that displaying works on jupyter-sphinx and sphinx-gallery at
 # the same time. Using PYGMT_USE_EXTERNAL_DISPLAY="false" in the Makefile
 # for sphinx-gallery to work means that fig.show won't display anything here
 # either.
 pygmt.set_display(method="notebook")
```

```python
import pygmt

fig = pygmt.Figure()
fig.grdimage(
    gravity,
    projection="W20c",
    cmap="viridis",
    shading="+a45+nt0.2",
)
fig.basemap(frame=["af", "WEsn"])
fig.colorbar(
    position="JCB+w10c",
    frame=["af", 'y+lmGal', 'x+lObserved gravity'],
)
fig.coast(shorelines=True, resolution="c", area_thresh=1e4)
fig.show()
```

```python
import boule as bl

ellipsoid = bl.WGS84
normal_gravity = ellipsoid.normal_gravity((gravity.longitude, gravity.latitude, gravity.height))
```

```python
fig = pygmt.Figure()
fig.grdimage(
    normal_gravity,
    projection="W20c",
    cmap="viridis",
    shading="+a45+nt0.2",
)
fig.basemap(frame=["af", "WEsn"])
fig.colorbar(
    position="JCB+w10c",
    frame=["af", 'y+lmGal', 'x+lNormal gravity'],
)
fig.coast(shorelines=True, resolution="c", area_thresh=1e4)
fig.show()
```

```python
gravity_disturbance = gravity - normal_gravity
gravity_disturbance
```

```python
import verde as vd

maxabs = vd.maxabs(gravity_disturbance, percentile=98)

fig = pygmt.Figure()
pygmt.makecpt(series=[-maxabs, maxabs], cmap="balance+h0", background=True)
fig.grdimage(
    gravity_disturbance,
    projection="W20c",
    cmap=True,
    shading="+a45+nt0.2",
)
fig.basemap(frame=["af", "WEsn"])
fig.colorbar(
    position="JCB+w10c+e",
    frame=["a20f10", 'y+lmGal', 'x+lGravity disturbance'],
)
fig.coast(shorelines=True, resolution="c", area_thresh=1e4)
fig.show()
```
